```python
from datetime import datetime
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Laden der Daten
df = pd.read_csv('../Neuronales_Netz/Fahrradzaehlung_komplett_rolling_mean_int.csv')
df['time'] = pd.to_datetime(df['time'], format='%Y-%m-%d %H:%M:%S')
df = df.set_index('time')
#df = df.resample('d').sum()
```

```python
start_date = datetime(2019, 12, 31)
end_date = datetime(2023, 5, 5)
plt.plot(df['estermannufer'], label = 'Estermannufer', color = 'blue')
plt.plot(df['estermannufer'][start_date:end_date], label = 'Estermannufer COVID-19', color = 'orange')
plt.title('Überquerungen während COVID-19 Estermannufer')
plt.xlabel('Datum')
plt.ylabel('Fahrradzaehlung')
plt.legend()
plt.show()
```
