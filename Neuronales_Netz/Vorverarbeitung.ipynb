```python
# Berechnung des Weekly_Mean
bruecken = df_gesamt.columns.values
time_train_gesamt = df_gesamt.index
X1 = df_gesamt[bruecken].values
week_means1 = pd.DataFrame(X1, index=time_train_gesamt, columns=bruecken)
week_means1 = week_means1.groupby(week_means1.index.isocalendar().week)[week_means1.columns].mean()
df_weekly_mean = pd.DataFrame(week_means1)
df_weekly_mean
```

```python
# Rolling Mean durchführen
woche=[]
for i in range(52):
    tmp = df_weekly_mean[bruecken][df_weekly_mean.index==i+1].values.tolist()
    woche.append(tmp)

for i in range(52):
    for j in range(len(bruecken)):
        start = df_gesamt[bruecken][df_gesamt.index.isocalendar().week==i+1].index
        anfang = start[0]
        ende = start[-1]
        df_gesamt[bruecken[j]][anfang:ende].fillna(value=woche[i][0][j], inplace=True)
```
