```python
# Daten laden 2022
df = pd.read_csv("../Projektdatensaetze_angepasst/2022/FahrradZaehlungBonn2022.csv")


# time Eigenschaften geben
df['time'] = pd.to_datetime(df['time'])
df['time'] = pd.to_datetime(df['time'], format = '%Y-%m-%d %H:%M:%S')


# Index setzen
df = df.set_index('time')

# auf Stunden umrechnen
df = df.resample('h').sum()

# addition der Werte
df['Bruehler Straße'] = df['Bruehler Straße Ri. Sueden'] + df['Bruehler Straße Ri. Norden'] + df['Bruehler Straße [Bike IN]'] + df['Brueler Straße [Bike OUT]']
df['Estermannufer'] = df['Estermannufer Ri. Bonn'] + df['Estermannufer Ri. Koeln'] + df['Estermannufer [Bike IN]'] + df['Estermannufer [Bike OUT]']
df['Kennedybruecke'] = df['Kennedybruecke Ri. Beuel'] + df['Kennedybruecke Ri. Innenstadt']
df['Nordbruecke'] = df['Nordbruecke Ri. Auerberg'] + df['Nordbruecke Ri. Beuel']
df['Suedbruecke'] = df['Suedbruecke Ri. Oberkassel'] + df['Suedbruecke Ri. Rheinaue']
df['Rhenusallee'] = df['Rhenusallee Ri. Beuel'] + df['Rhenusallee Ri. Oberkassel'] + df['Rhenusallee [Bike IN]'] + df['Rhenusallee [Bike IN].1'] + df['Rhenusallee [Bike OUT]'] + df['Rhenusallee [Bike OUT].1']
df['Von-Sandt-Ufer'] = df['Von-Sandt-Ufer Ri. Bad Godesberg'] + df['Von-Sandt-Ufer Ri. Bonn'] + df['Von-Sandt-Ufer [Bike IN]'] + df['Von-Sandt-Ufer [Bike IN].1'] + df['Von-Sandt-Ufer [Bike OUT]'] + df['Von-Sandt-Ufer [Bike OUT].1']
df['Weg auf Damm Neil'] = df['Weg auf Damm Neil Ri. Norden'] + df['Weg auf Damm Neil Ri. Sueden'] + df['Weg auf Damm Neil Ri. Norden.1'] + df['Weg auf Damm Neil Ri. Sueden.1']
df['Wilhelm-Spiritus-Ufer'] = df['Wilhelm-Spiritus-Ufer Ri. Sueden']  + df['Wilhelm-Spiritus-Ufer Ri. Sueden.1'] + df['Wilhelm-Spiritus-Ufer Ri. Norden'] + df['Wilhelm-Spiritus-Ufer Ri. Sueden.1']

# in Liste speichern
zeitraum = zeitraum + df.index.tolist()
bruehler_straße = bruehler_straße + df[('Bruehler Straße')].values.tolist()
estermannufer = estermannufer + df[('Estermannufer')].values.tolist()
kennedybruecke = kennedybruecke + df[('Kennedybruecke')].values.tolist()
nordbruecke = nordbruecke + df[('Nordbruecke')].values.tolist()
suedbruecke = suedbruecke + df[('Suedbruecke')].values.tolist()
rhenusallee = rhenusallee + df[('Rhenusallee')].values.tolist()
von_sandt_ufer = von_sandt_ufer + df[('Von-Sandt-Ufer')].values.tolist()
weg_auf_dem_damm = weg_auf_dem_damm + df[('Weg auf Damm Neil')].values.tolist()
wilhelm_spiritus_ufer = wilhelm_spiritus_ufer + df[('Wilhelm-Spiritus-Ufer')].values.tolist()
```

```python
# Ausgabe der Listen um auf Länge zu testen 2022
print(len(zeitraum))
print(len(bruehler_straße))
print(len(estermannufer))
print(len(kennedybruecke))
print(len(nordbruecke))
print(len(suedbruecke))
print(len(rhenusallee))
print(len(von_sandt_ufer))
print(len(weg_auf_dem_damm))
print(len(wilhelm_spiritus_ufer))
```
